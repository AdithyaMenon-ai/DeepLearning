```python
import mediapipe as mp

print(mp.__version__)
print(mp.solutions)
```

```python
mp_hand = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
```

```python
data = mp_hand.Hands()
```

```python
Video = cv2.VideoCapture(0)
while True:
    suc, img = Video.read()

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    result = data.process(img_rgb)
    print(result.multi_hand_landmarks)

    if result.multi_hand_landmarks:
        for x in result.multi_hand_landmarks:
            mp_drawing.draw_landmarks(img, x, mp_hand.HAND_CONNECTIONS,mp_drawing.DrawingSpec(color=(0, 0, 0),thickness=5,circle_radius=2),mp_drawing.DrawingSpec(color=(0, 255, 191),thickness=2,circle_radius=2))

    cv2.imshow("Hand", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

Video.release()
cv2.destroyAllWindows()
```
