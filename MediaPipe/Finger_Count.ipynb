```python
import cv2
import mediapipe as mp
```

```python
mp_hand = mp.solutions.hands
mp_draw = mp.solutions.drawing_utils
```

```python
data = mp_hand.Hands()
```

```python
Video = cv2.VideoCapture(0)

while True:
    suc, img = Video.read()
    img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    result = data.process(img1)
    
    tipid = [4,8,12,16,20]
    lmlist=[]


    if result.multi_hand_landmarks:
        for x in result.multi_hand_landmarks:
            for id,lm in enumerate(x.landmark):
                cx = lm.x
                cy = lm.y
                lmlist.append([id,cx,cy])

                if len(lmlist)!=0 and len(lmlist)==21:
                    finger_count = []
                    
                    for i in range(5):
                        if lmlist[tipid[i]][2]>lmlist[tipid[i]-2][2]:
                            finger_count.append(0)
                        else:
                            finger_count.append(1)
                    print(finger_count)

            mp_draw.draw_landmarks(img, x, mp_hand.HAND_CONNECTIONS)

        



    cv2.imshow("Finger Counter", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

Video.release()
cv2.destroyAllWindows()
```
