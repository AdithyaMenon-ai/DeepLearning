```python
import cv2
import mediapipe as mp
```

```python
mp_face = mp.solutions.face_mesh
mp_drawing = mp.solutions.drawing_utils
mp_styles = mp.solutions.drawing_styles
```

```python
data = mp_face.FaceMesh()
```

```python
Video = cv2.VideoCapture(0)
while True:
    suc, img = Video.read()
    img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    result = data.process(img1)

    if result.multi_face_landmarks:
        for x in result.multi_face_landmarks:
            mp_drawing.draw_landmarks(img, x, mp_face.FACEMESH_TESSELATION, connection_drawing_spec=mp_styles.get_default_face_mesh_tesselation_style())

    cv2.imshow("Face Landmarks", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

Video.release()
cv2.destroyAllWindows()
```
