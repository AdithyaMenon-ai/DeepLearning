```python
import cv2
import matplotlib.pyplot as plt
```

```python
face=cv2.CascadeClassifier('haarcascade_eye.xml')
```

```python
# #video

import cv2

video = cv2.VideoCapture('test_video.mp4')  # webcam # video=cv2.VideoCapture(0)#0 means webcam,..1 means external webcam,...or video path 
while True:
    success, img = video.read()
    if success:
        img1=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
        faces=face.detectMultiScale(img1,minNeighbors=10)
        
        for (x,y,w,h) in faces:
            cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)

        cv2.imshow("video", img)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
video.release()
cv2.destroyAllWindows()
```
