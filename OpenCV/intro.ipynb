```python
#TO Binary,....only after grey_scale

#Thresh_Binary
thresh,bin_img=cv2.threshold(img_grey,50,200,cv2.THRESH_BINARY)#,.....first 50 pix will be black other to white
bin_img
```

```python
cv2.imshow('binary_image',bin_img)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
#Thresh_OTSU,...using this machine will default give the values

thresh,bin_img1=cv2.threshold(img_grey,100,255,cv2.THRESH_OTSU)
bin_img1
```

```python
#To Display the image,...matplotlib
import matplotlib.pyplot as plt
img=cv2.imread('baby_img_1.png')
img
```

```python
plt.imshow(img)
```

y values is generating highest to lowest
its a BGR image


#RGB
#CV2.READ ====> BGR

#Display ====> cv2 ====  it again convert back to RGB

but using display matplotlib,...===> it will read as BGR

```python
#Convert BGR to RGB
conv_img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(conv_img)
```

```python
#Bounding Box
    #rectangle
    #circle
```

```python
#rectangle
    #left_top & right_bottom
cv2.rectangle(conv_img,pt1=(220,40),pt2=(390,225),color=(255,0,0),thickness=3)
plt.imshow(conv_img)
```

```python
img=cv2.imread('baby_img_2.png')
conv_img_2 = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
cv2.rectangle(conv_img_2,pt1=(280,200),pt2=(460,420),color=(255,0,0),thickness=3)
plt.imshow(conv_img_2)
```

```python
img=cv2.imread('baby_img_2.png')
conv_img_2_1 = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
cv2.circle(conv_img_2_1,center=(390,320),radius=130,color=(255,0,0),thickness=3)#thickness -ve values make the portion covered
plt.imshow(conv_img_2_1)
```

```python
#Text

img1=cv2.imread('baby_img_1.png')
conv_img_1_1=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
plt.imshow(conv_img_1_1)
```

```python
#Cute Baby ====> TEXT
        #(PUTTEXT)

cv2.putText(conv_img_1_1,text='cute_baby',org=(200,500),fontFace=cv2.FONT_HERSHEY_COMPLEX,fontScale=1,color=(255,0,0),thickness=1)
plt.imshow(conv_img_1_1)
```

```python
img2_2=cv2.imread('baby_img_2.png')
conv_img_2_2=cv2.cvtColor(img2_2,cv2.COLOR_BGR2RGB)
cv2.rectangle(conv_img_2_2,pt1=(280,200),pt2=(460,420),color=(255,0,0),thickness=3)
cv2.putText(conv_img_2_2,text='cute_baby',org=(200,500),fontFace=cv2.FONT_HERSHEY_COMPLEX,fontScale=1,color=(255,0,0),thickness=1)
plt.imshow(conv_img_2_2)
```
