```python
import speech_recognition as sr
```

```python
def speech_txt():
    r=sr.Recognizer()
    try:
        while True:

            with sr.Microphone() as source:
                audio = r.listen(source)
            try:
                print(r.recognize_google(audio))
            except:
                print("Didn't hear anything,...pls repeat")

    except KeyboardInterrupt:
        print("Stopped by user")

speech_txt()
```
