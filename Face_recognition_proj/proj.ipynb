```python
#loop it for all the images
def face_encodings(images):
    encodinglist=[]
    for img in imgs:
        img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
        face_in_frame=fr.face_locations(img)
        #print face in frames
        face_encoder=fr.face_encodings(img,face_in_frame)[0]
        #print face encodings
        encodinglist.append(face_encoder)
    return encodinglist
```

```python
encoderlist_knownfaces=face_encodings(imgs)
print(encoderlist_knownfaces)
```

```python
Video=cv2.VideoCapture(0)
while True:
    suc,frame = Video.read()
    if not suc or frame is None:
        print("⚠️ Frame not received, exiting...")
        break
    frame1 = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
    face_in_frame=fr.face_locations(frame1)
    #print(face_in_frame)
    encode_face=fr.face_encodings(frame1,face_in_frame)
    #print(encode_face)
    for enc_face,faceloc in zip(encode_face,face_in_frame):
        matches = fr.compare_faces(encoderlist_knownfaces,enc_face)
        print(matches)


    cv2.imshow('Face',frame)
    if cv2.waitKey(1)&0XFF==ord('q'):
        break
    Video.release()
    cv2.destroyAllWindows()
```
